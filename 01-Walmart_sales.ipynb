## Preprocessing

Nous devons d'abord supprimer les valeurs cibles non renseignées. 

```python
df = df.dropna(subset=["Weekly_Sales"])
print("Lignes restantes :", df.shape[0])
```

Nous devons également nous intéresser aux lignes sans date. Celles-ci sont moins évidentes à trancher. Nous devons les décomposer en quatre variables. Year, Month, Day et DayOfWeek.
Voyons d'abord quelles sont les lignes comportant une date absente. 

```python
df[df["Date"].isna()]
```

La majeure partie des lignes porte des informations valides et seule la date manque. Imputer quatre variables temporelles, reviendrais à inventer un mois et une année. Or, le mois est justement ce qui est censé capter l'effet de fin d'année observé plus haut. Lui donner des valeurs fabriquées pourrait degrader la seule variable temporelle que nous avons pu observer à être utile. 
Imputer une date est plus conséquent que d'imputer une valeur susceptible de varier autour d'une moyenne, telle qu'une température ou un prix. Le choix est donc fait de ne pas retenir ces 18 lignes-là, en dépit du faible volume d'information à disposition dans le dataset. 


```python
df = df.dropna(subset=["Date"])
print("Lignes restantes :", df.shape[0])
```

Le jeu de données tombe à 118 lignes. 

#### conversion des dates
L'enoncé impose de convertir les dates en 4 variables.

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["DayOfWeek"] = df["Date"].dt.dayofweek

df[["Date", "Year", "Month", "Day", "DayOfWeek"]].head(10)
```

On constate que DayOfWeek est systematiquement egal à 4, soit un vendredi (lundi = 0). On peut le verifier sur l'ensemble des jours:

```python
df["DayOfWeek"].value_counts()
```

Cette colone peut donc etre supprimee.

```python
df = df.drop(columns=["DayOfWeek"])
```

#### Filtrage à 3 sigma

```python
colonnes = ["Temperature", "Fuel_Price", "CPI", "Unemployment"]

for col in colonnes:
    moyenne = df[col].mean()
    ecart_type = df[col].std()
    df = df[(df[col] >= moyenne - 3 * ecart_type) & (df[col] <= moyenne + 3 * ecart_type) | df[col].isna()]

print("Lignes restantes :", df.shape[0])
```

Les cinq lignes du magasin numéro 12 que nous avons les étudiés précédemment et dont le taux de chomage était un outlier ont été supprimés. Le nombre de lignes restantes passe donc de 118 à 113. 

#### Séparation des variables explicatives et de la cible

```python
Y = df["Weekly_Sales"]
X = df.drop(columns=["Weekly_Sales", "Date"])

X.head()
```

Y est la cible. On retire également les dates, car elles sont déjà incluses dans les 3 variables qu'on vient de définir: Year, Month, et Day. 

#### Découpage train / test

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=0)

print("Train :", X_train.shape[0], "lignes")
print("Test  :", X_test.shape[0], "lignes")
```

#### Pipeline de preprocessing

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_features = ["Temperature", "Fuel_Price", "CPI", "Unemployment", "Year", "Month", "Day"]
categorical_features = ["Store", "Holiday_Flag"]

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler()),
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(drop="first")),
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features),
])
```

```python
X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)

print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)
```

Les 8 variables de départ donnent 26 colonnes : 7 numériques, 1 indicatrice pour Holiday_Flag et 18 pour Store.

## Training

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, Y_train)

print("R² train :", model.score(X_train, Y_train))
print("R² test  :", model.score(X_test, Y_test))
```

L'écart entre les deux valeurs du R² reste faible et le surapprentissage modéré. 
À ce stade, le modèle explique 94 % de la variance des ventes sur des données qu'il n'a jamais vues. 

```python
feature_names = numeric_features + list(
    preprocessor.named_transformers_["cat"]
    .named_steps["encoder"]
    .get_feature_names_out(categorical_features)
)

coefficients = pd.DataFrame({
    "feature": feature_names,
    "coefficient": model.coef_,
}).sort_values("coefficient")

px.bar(coefficients, x="coefficient", y="feature", orientation="h",
       title="Coefficients du modèle de référence", height=700)
```

On observe que les coefficients associés à chaque magasin sont nettement prépondérants face à toute autre variable. L'intégralité des variables mesurées autres que le store, telles que notamment le CPI, le unemployment, holiday_flag, fuel_price, day, temperature, year ou months, sont toutes localisées autour du zéro.

#### Lasso avec GridSearchCV

Le modele de référence compte 26 colonnes avec 90 lignes d'entrainement, 18 indiques le magasin estimées sur quelques observations chacune. On passe donc a un Lasso, dont l'alpha est réglé par GridSearchCV.

```python
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV

params = {"alpha": [ 0.1, 1.0, 10.0, 100.0, 1000.0, 10000.0, 50000]}

gridsearch = GridSearchCV(Lasso(max_iter=10000), param_grid=params, cv=5)
gridsearch.fit(X_train, Y_train)

print("Meilleur alpha :", gridsearch.best_params_)
print("Meilleur score en validation croisée :", gridsearch.best_score_)
print("R² train :", gridsearch.score(X_train, Y_train))
print("R² test  :", gridsearch.score(X_test, Y_test))
```

Le meilleur alpha ressort a 1000.

Le R² de test passe de 0,940 a 0,956, tandis que celui du train baisse légerement, de 0,973 a 0,971. La régularisation a donc atteint son but : le modele perd un peu de précision sur les données qu'il connait, et en gagne sur celles qu'il n'a jamais vues. L'écart entre train et test tombe de 3,3 a 1,5 point 

Lasso a la particularité d'annuler completement les coefficients des variables qu'il juge inutiles. On regarde donc lesquelles ont été écartées.

```python
lasso_coefficients = pd.DataFrame({
    "feature": feature_names,
    "coefficient": gridsearch.best_estimator_.coef_,
}).sort_values("coefficient")

print("Coefficients annulés :", (lasso_coefficients["coefficient"] == 0).sum(), "sur", len(feature_names))

px.bar(lasso_coefficients, x="coefficient", y="feature", orientation="h",
       title="Coefficients du Lasso", height=700)
```

Deux coefficients sont annulés : Holiday_Flag et Year. Ces annulations confirment l'analyse exploratoire. 

Holiday_Flag marque la semaine contenant Noël alors que le pic de ventes se produit dans les semaines précédentes : le drapeau ne coincidait pas avec l'effet, le modele l'écarte. Year était sans effet visible sur le nuage de points, il est egalement ecarté.

Month est en revanche conservé, avec un coefficient supérieur a toutes les variables économiques : c'est lui qui capte entre autres l'effet de fin d'année.

## Conclusion

Le résultat a été obtenu pour un coefficient alpha = 1 000. Ce coefficient est probablement optimisable quelque part autour de cette valeur, peut-être 2 000 ou 3 000 pour un gain probablement très faible.
Avec ce modèle, nous obtenons les résultats suivants : 
R² train : 0.97

Ces résultats constituent une amélioration par rapport au modèle sans Lasso. Et Lasso s'avère généraliser mieux en écartant deux variables sans effet.

##### Limitation du modele

Le modèle est toutefois limité en analyse dans la mesure où seules 113 lignes ont été exploitées dont 23 pour le test.
Pourra aussi noter que le filtrage à 3 sigma a été appliqué avant le découpage train-test. Les seuils ont donc été calculés en incluant les lignes de test, ce qui constitue une fuite de données. Filtrer après le découpage aurait été plus rigoureux.
